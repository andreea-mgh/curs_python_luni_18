# Limbaje de programare compilate vs interpretate

Compilat -> tot codul se transforma in machine code, iar daca codul e incorect atunci nu se poate transforma si NU putem rula cod incorect

Avantaje: MULT mai rapid decat codul interpretat

Dezavantaje: mai greu de scris, mai putin flexibil

Interpretat -> codul se converteste la cod masina in memoria calculatorului, linie cu linie, si putem rula codul chiar daca nu e tot corect

Avantaje: usor de scris, flexibil, usor de testat

Dezavantaje: mai incete (uneori chair de 10 ori mai incete)

# Formatarea codului: case sensitivity si indentare

Case sensitivity -> cand "pisoi" e diferit de "Pisoi"

Majoritatea limbajelor de programare sunt case-sensitive. Multe componente din sistemul Windows, de exemplu, sunt case-insensitive.

Indentare -> spatiul care se adauga pentru a deosebi instructiuni de alte instructiuni din blocuri de cod diferite.

# Tipuri de date

Tipurile de date sunt formele prin care putem stoca informatie in memoria calculatorului.

Tipurile comune sunt:

* Numar
  * INT (numar intreg): 4, 235189, -3, -2000, 0
  * FLOAT (numar cu virgula): 1.5, 3.14, -0.2, 0.01, 4.0
* String (text): "abc", 'Laura', "00000", "21", "123abc", "0"
* Boolean: True, False
* None

```python
print(type(2))
print(type(3.14))
print(type("3.14"))
print(type(True))
print(type(None))
```
